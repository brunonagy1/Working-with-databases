```python
def read_only_query(db:str,sql_query:str) -> pd.DataFrame | None:
    """Funkcija koja kreira SQL upit za bazu podataka db. Ako se funkcija uspjesno izvrsi output je pd.DataFrame, a inace None."""
    try:
        with sqlite3.connect(db) as conn:
            return pd.read_sql(sql_query,conn)
    except Exception as e:
        print(e)
        return None
```

```python
read_only_query("database2.db","SELECT * FROM customers")
```

```python
read_only_query("database2.db","SELECT Name,City FROM customers WHERE City LIKE 'Z%'")
```

```python
read_only_query("database2.db","SELECT Name, City FROM customers WHERE Name LIKE 'P_%'")
```
