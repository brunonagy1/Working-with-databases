Kupci koji su napravili barem jednu narudžbu

```python
sqldf("SELECT c.Name FROM customers_df c JOIN orders_df o ON c.CustomerID=o.CustomerID")
```

Prikaži sve narudžbe zajedno s cijenom proizvoda

```python
sqldf("SELECT o.Product, SUM(o.Quantity*p.Price) FROM orders_df o JOIN products_df p ON o.Product=p.Product GROUP BY p.Product")
```

Pronađi ukupnu količinu naručenih proizvoda po kupcu

```python
sqldf("SELECT c.Name, SUM(o.Quantity) FROM customers_df c JOIN orders_df o ON c.CustomerID=o.CustomerID GROUP BY c.Name")
```

Prikaži sve narudžbe zajedno s cijenom proizvoda i ukupnom vrijednošću narudžbe

```python
sqldf("SELECT o.Product, p.Price,  SUM(o.Quantity*p.Price) AS total FROM orders_df o JOIN products_df p ON o.Product=p.Product GROUP BY o.Product")
```

Prikaži proizvode koji su naručeni više od jednom

```python
sqldf("SELECT o.Product FROM orders_df o JOIN products_df p ON o.Product=p.Product WHERE o.Quantity>1")
```

Prikaži sve kupce koji nikada nisu naručili Tablet

```python
sqldf("SELECT Name FROM customers_df WHERE Name NOT IN (SELECT c.Name FROM customers_df c JOIN orders_df o ON c.CustomerID=o.CustomerID WHERE o.Product='Tablet')")
```
